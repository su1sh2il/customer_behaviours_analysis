### Replacing null value with median of their Review Rating along with category

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

### Convert all column into lower cases with insted of sapace use undesrscore

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns = {'purchase_amount_(usd)':'purchase_amount'})
```

### Creating new column age_group

```python
labels = ['Young Adult','Adult','Middle_Senior','Senior']
df['age_group'] = pd.qcut(df['age'], q = 4, labels = labels)
```

```python
df[['age', 'age_group']].head()
```

```python
df['frequency_of_purchases'].unique()
```

### Create a column Frequency_purchase_day

```python
fequency_mapping = {
    'Fortnightly':14,
    'Weekly':7,
    'Annually':365,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Monthly':30,
    'Every 3 Months':90
}
df['frequency_purchases_day'] = df['frequency_of_purchases'].map(fequency_mapping)
```

```python
df[['frequency_of_purchases','frequency_purchases_day']].head()
```

### Drop promo_code_used because it is redundant

```python
# gives TRUE means has redundant column we should removed that
(df['discount_applied'] == df['promo_code_used']).all()
```
